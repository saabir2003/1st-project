```python
class BankAccount:
    def __init__(self,i,name,balance):
        self.i=i
        self.name=name
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print("deposited suceesfully")
        self.printbalance()
    def withdraw(self,amount):
        if self.balance<amount:
            print("balance insufficent")
        else:
            self.balance-=amount
            print("amount withdrawn")
            self.printbalance()
    def printbalance(self):
        print(self.balance)
b1=BankAccount(1,"saabir",10000)
b2=BankAccount(1,"ayesha",13000)
```

```python
b1.deposit(10000)
```

```python
# inheritance in python
# single class inheritance
class transaction(BankAccount):
    def __init__(self,i,name,balance,amount,credit):
        super().__init__(i,name,balance)
        self.amount=amount
        self.credit=credit
    def printl(self):
        print(self.i,self.balance,self.amount,self.credit,self.name)
```

```python
class data:
    def __init__(self,dname,data1):
        self.dname=dname
        self.data1=data1
    def printdata():
        print(self.dname,self.data1)
```

```python
#multipple inhertiance
# when a class nherits more than one base class
class Item(data):
    def __init__(self,name,Price,Size,data1,dname):
        data.__init__(self,dname,data1)
        self.name=name
        self.Price=Price
        self.Size=Size
    def printdetails(self):
        print(self.name,self.Price,self.Size)
I1=Item("butter",20,"M",30,"KB")
I1.printdetails()
```

```python
class customer(data):
    def __init__(self,cname,dname,data1):
        data.__init__(self,dname,data1)
        self.cname=cname
    def printcname(self):
        print(self.cname)
C1=customer("saabir","MB",45)
C1.printcname()
```

```python
class Order(Item,customer):
    def __init__(self,name,Price,Size,cname,order_id,data1,dname):
        Item.__init__(self,name,Price,Size,data1,dname)
        customer.__init__(self,cname,data1,dname)
        #data.__init__(self,data1,dname)
        self.order_id=order_id
    def printdetails(self):
        print(self.name,self.Price,self.Size,self.order_id,self.cname,self.data1,self.dname)
O1=Order("butter",20,"M","saabir",23,67,"GB")
```

```python
class shape:
    def area(self):
        print("area of a shape")
class square(shape):
    def __init__(self,side):
        self.side=side
    def area(self):
        print(self.side*self.side)
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        print(self.radius*self.radius*math.pi)
s1=square(23)
c1=circle(3)
#method overriding
```

```python
from abc import ABC,abstractmethod
# define a abstract class
class vehicle(ABC):
    @abstractmethod
    def startengine(self):
        #since it is abstract class it needs to be empty
        pass
class car(vehicle):
    # abstract method enforces the commom methods if the commom methods are not implemented then it shows an error
    #while creating an object
    def startengine(self):
        print("start engine")
```

```python
class Person:
    def __init__(self,name,age,gender):
        self._name=name# protected variables
        self._age=age# protected variables it cannot be access outside the class except the derived class
        self.gender=gender#public variable
P1=Person("ayesha",23,"Female")
# we are retricting some of the attributes in the class
class student(Person):
    def __init__(self,name,age,gender,year):
        Person.__init__(self,name,age,gender)
        self.year=year
    def print(self):
        print(self._name,self._age,self.gender,self.year)
s1=student("arfa",45,"Male",2003)
s1.print()
#s1._name="arfa1"

#dir(P1)
```

```python
@functio
def add():
    print("this is another function in add")
# this function will go in as a parameter
# decorators are used to modity or add functionality 
#without changing the existing code
```

```python
def function1(func):
    def wapper():
        print("something before the function")
        func()
        print("something after the function")
    return wapper()

@function1# this is decorator function where the add
#function goes in as a parameter
def add(a=3,d=5):
    print(a+d)
```
