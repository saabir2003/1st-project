```python
class BankAccount:
    def __init__(self,i,name,balance):
        self.i=i
        self.name=name
        self.balance=balance
    def deposit(self,amount):
        self.balance+=amount
        print("deposited suceesfully")
        self.printbalance()
    def withdraw(self,amount):
        if self.balance<amount:
            print("balance insufficent")
        else:
            self.balance-=amount
            print("amount withdrawn")
            self.printbalance()
    def printbalance(self):
        print(self.balance)
b1=BankAccount(1,"saabir",10000)
b2=BankAccount(1,"ayesha",13000)
```

```python
b1.deposit(10000)
```
